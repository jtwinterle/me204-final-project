## Connecting to the Open Brewery DB API

This project collects brewery information using the Open Brewery DB API. The API was selected because it is publicly available, requires no authentication, and returns structured JSON data that can be collected programmatically. These characteristics make the data collection process transparent and reproducible.

The API returns brewery information in pages, so multiple requests will be required to build a dataset large enough for meaningful analysis.

```python
# Open Brewery DB API endpoint

BASE_URL = "https://api.openbrewerydb.org/v1/breweries"

PER_PAGE = 200
MAX_PAGES = 10

print("API configured successfully.")
```

### Why these settings?

The API allows up to 200 brewery records to be returned in a single request. Using the maximum page size reduces the total number of requests while ensuring the collection process remains efficient. Ten pages are collected initially, producing a dataset of up to 2,000 brewery records, which is sufficient for the analyses planned in this project.

## Collect Brewery Data

The Open Brewery DB API returns brewery records in pages rather than all at once. To create a dataset large enough for meaningful analysis, this notebook requests multiple pages of data and combines the results into a single collection.

The original API responses are preserved exactly as returned. No cleaning or filtering is performed during this stage because all preparation will take place in NB02. This separation ensures that the raw data remains available for verification and reproducibility.

```python
# Download brewery data from multiple API pages

all_breweries = []

for page in range(1, MAX_PAGES + 1):
    response = requests.get(
        BASE_URL,
        params={
            "page": page,
            "per_page": PER_PAGE
        }
    )

    response.raise_for_status()

    breweries = response.json()

    if not breweries:
        break

    all_breweries.extend(breweries)

    print(f"Collected page {page} ({len(breweries)} breweries)")

print(f"\nTotal breweries collected: {len(all_breweries)}")
```

### Discussion

The data collection process completed successfully, and brewery records were downloaded directly from the Open Brewery DB API. Collecting multiple pages increased the size of the dataset while preserving the original API responses exactly as returned.

At this stage, no records have been modified, removed, or cleaned. Preserving the original data ensures that every transformation performed in later notebooks can be reproduced and verified against the raw source.

```python
# Save the raw API response

raw_file = Path("../data/raw/breweries_raw.json")

with open(raw_file, "w") as f:
    json.dump(all_breweries, f, indent=2)

print(f"Raw data saved to: {raw_file}")
```

```python
# Verify that the raw data file was created

print(raw_file.exists())
print(f"Number of brewery records: {len(all_breweries)}")
```
